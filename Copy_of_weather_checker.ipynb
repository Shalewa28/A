```
import requests

# --------------------------------------------
# Weather Checker Application
# --------------------------------------------
# This program allows a user to enter the name
# of a city and retrieve real-time weather data
# using an external weather API.
#
# It displays:
# - Temperature
# - Weather conditions
# - Humidity
#
# Author: Akegbesola Mercy
# Course: Python Programming
# --------------------------------------------


# API key from OpenWeatherMap
API_KEY = "2dd2cb062cfb83017ed08e7f2df64ddd"


def get_weather(city):
    """
    Connects to the OpenWeatherMap API and retrieves
    weather data for a given city.

    Parameters:
        city (str): Name of the city entered by the user

    Returns:
        dict: Weather data if successful
        None: If city is invalid or request fails
    """

    url = "https://api.openweathermap.org/data/2.5/weather"

    # Parameters sent to the API
    params = {
        "q": city,
        "appid": API_KEY,
        "units": "metric"  # Celsius temperature
    }

    try:
        response = requests.get(url, params=params)
        data = response.json()

        # If API returns error code, city was not found
        if data["cod"] != 200:
            return None

        return data

    except Exception:
        # Handles network errors or request failures
        return None


def display_weather(data):
    """
    Extracts and displays weather information
    in a clean, readable format.

    Parameters:
        data (dict): Weather data from API
    """

    # Extracting information from API response
    city_name = data["name"]
    country = data["sys"]["country"]

    temperature = data["main"]["temp"]
    feels_like = data["main"]["feels_like"]
    humidity = data["main"]["humidity"]

    weather_description = data["weather"][0]["description"]

    # Display output to user
    print("\n🌤️ Weather Information 🌤️")
    print("----------------------------")
    print(f"Location: {city_name}, {country}")
    print(f"Temperature: {temperature}°C")
    print(f"Feels Like: {feels_like}°C")
    print(f"Condition: {weather_description.title()}")
    print(f"Humidity: {humidity}%")
    print("----------------------------\n")


def main():
    """
    Main function that runs the program.
    Handles user interaction and program loop.
    """

    print("🌍 Welcome to the Weather Checker App 🌍")

    # Loop allows user to check multiple cities
    while True:
        city = input("\nEnter a city name: ")

        weather_data = get_weather(city)

        # Check if data was successfully retrieved
        if weather_data:
            display_weather(weather_data)
        else:
            print("❌ Error: City not found or unable to retrieve data.")

        # Ask user if they want to continue
        repeat = input("Would you like to check another city? (yes/no): ").lower()

        if repeat != "yes":
            print("\nThank you for using the Weather Checker App! 👋")
            break


# Runs the program
if __name__ == "__main__":
    main()
```
